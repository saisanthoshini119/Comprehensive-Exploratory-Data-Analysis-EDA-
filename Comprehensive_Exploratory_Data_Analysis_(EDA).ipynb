```python
df = pd.read_csv("tested.csv")

# Display first five rows
display(df.head())

# Display last five rows
display(df.tail())

# Dataset dimensions
print("Shape:", df.shape)

# Column names
print("Columns:", df.columns.tolist())

# Dataset information
df.info()

# Statistical summary
display(df.describe(include="all"))
```

```python
# Missing values
print("Missing values:")
print(df.isnull().sum())

# Missing value percentages
print("\nMissing value percentage:")
print((df.isnull().mean() * 100).round(2))

# Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing numerical values with the median
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with the mode
cat_cols = df.select_dtypes(include=["object", "category"]).columns

for col in cat_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Separate column types
numerical = df.select_dtypes(include=np.number)
categorical = df.select_dtypes(
    include=["object", "category", "bool"]
)

print("Numerical columns:", numerical.columns.tolist())
print("Categorical columns:", categorical.columns.tolist())

# Numerical summary
display(numerical.describe())

# Unique values in categorical columns
for col in categorical.columns:
    print(f"\n{col}:")
    print(df[col].value_counts().head(10))
```

```python
# Comprehensive Dataset Summary

print("=" * 50)
print("       DATASET SUMMARY")
print("=" * 50)

print("\n1. Dataset Shape")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n2. Column Names")
print(df.columns.tolist())

print("\n3. Dataset Information")
df.info()

print("\n4. Statistical Summary")
display(df.describe(include="all").T)

print("\n5. Missing Values")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0
      else "No missing values")

print("\n6. Duplicate Rows")
print(df.duplicated().sum())

print("\n7. Data Types")
print(df.dtypes)

print("\n8. Unique Values in Each Column")
print(df.nunique())

print("\n9. Numerical Columns")
print(df.select_dtypes(include="number").columns.tolist())

print("\n10. Categorical Columns")
print(df.select_dtypes(include=["object", "category"]).columns.tolist())

print("\n11. First Five Records")
display(df.head())

print("\n12. Final Summary")
print(f"Total Records: {df.shape[0]}")
print(f"Total Features: {df.shape[1]}")
print(f"Total Missing Values: {df.isnull().sum().sum()}")
print(f"Total Duplicate Rows: {df.duplicated().sum()}")
print("=" * 50)
```
